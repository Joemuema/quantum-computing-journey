Exercise 1 — Pauli gates

Start with |0⟩. Apply X, Y and Z separately. Print each resulting state and its measurement probabilities.

```python
import numpy as np

# Computational basis states
zero = np.array([[1], [0]], dtype=complex)

# Pauli gates
X = np.array([
    [0, 1],
    [1, 0]
], dtype=complex)

Y = np.array([
    [0, -1j],
    [1j, 0]
], dtype=complex)

Z = np.array([
    [1, 0],
    [0, -1]
], dtype=complex)

X_zero = X @ zero
print("X|0> =\n", X_zero)
print("Measurement of X|0> in computational basis: \n", np.abs(X_zero)**2)

Y_zero = Y @ zero
print("Y|0> =\n", Y_zero)
print("Measurement of Y|0> in computational basis: \n", np.abs(Y_zero)**2)

Z_zero = Z @ zero
print("Z|0> =\n", Z_zero)
print("Measurement of Z|0> in computational basis: \n", np.abs(Z_zero)**2)
```

Exercise 2 — Hadamard properties

Verify in Python that H @ H equals the identity matrix. Then apply H twice to |1⟩ and explain the result.

```python
# Hadamard
H = (1 / np.sqrt(2)) * np.array([
    [1, 1],
    [1, -1]
], dtype=complex)

identity_matrix = np.eye(2, dtype=complex)
result = H @ H

print(result)

one = np.array([[0], [1]], dtype=complex)
double_H_one = H @ (H @ one)

print("H H|1> =\n", double_H_one)
print("Measurement of H H|1> in computational basis: \n", np.abs(double_H_one)**2)
```

Exercise 3 — Phase and interference

Start with |0⟩, apply H, then Z, then H. Print the final state and measurement probabilities. Explain why the result differs from applying H twice.

```python
H_Z_H_zero = H @ (Z @ (H @ zero))
print("H Z H|0> =\n", H_Z_H_zero)
print("Measurement of H Z H|0> in computational basis: \n", np.abs(H_Z_H_zero)**2)

H_H_zero = H @ (H @ zero)

print("H H|0> =\n", H_H_zero)
print("Measurement of H H|0> in computational basis: \n", np.abs(H_H_zero)**2)
```

Exercise 4 — Rotation gates

Apply Ry(π/2) to |0⟩. Calculate the resulting state and measurement probabilities. Explain what happens.

```python
def Ry(theta):
    return np.array([
        [np.cos(theta/2), -np.sin(theta/2)],
        [np.sin(theta/2), np.cos(theta/2)]
    ], dtype=complex)

rotation_angle = np.pi / 2
Ry_gate = Ry(rotation_angle)
print("Ry gate with angle π/2:\n", Ry_gate)

print("Ry|0> =\n", Ry_gate @ zero)
print("Measurement of Ry|0> in computational basis: \n", np.abs(Ry_gate @ zero)**2)
```

Exercise 5 — Circuit order

Calculate X @ H @ |0⟩ and H @ X @ |0⟩. Explain why the two results differ.

```python
X_H_zero = X @ (H @ zero)
print("X H|0> =\n", X_H_zero)
print("Measurement of X H|0> in computational basis: \n", np.abs(X_H_zero)**2)

H_X_zero = H @ (X @ zero)
print("H X|0> =\n", H_X_zero)
print("Measurement of H X|0> in computational basis: \n", np.abs(H_X_zero)**2)
```

Exercise 6 — Build a reusable circuit

Write a function that accepts an initial state and a list of gates, applies them in sequence, and returns the final state. Test it with at least three different circuits.

```python
# Phase gates
S = np.array([
    [1, 0],
    [0, 1j]
], dtype=complex)

T = np.array([
    [1, 0],
    [0, np.exp(1j * np.pi / 4)]
], dtype=complex)

def Rx(theta):
    return np.array([
        [np.cos(theta/2), -1j*np.sin(theta/2)],
        [-1j*np.sin(theta/2), np.cos(theta/2)]
    ], dtype=complex)

def Rz(theta):
    return np.array([
        [np.exp(-1j*theta/2), 0],
        [0, np.exp(1j*theta/2)]
    ], dtype=complex)


def circuit_sequence(initial_state, gates):
    state = initial_state
    for gate in gates:
        state = gate @ state
    return state

S_H_Rz_H_zero = circuit_sequence(zero, [H, Rz(np.pi/4), H, S])
print("S H Rz(π/4) H|0> =\n", S_H_Rz_H_zero)
print("Measurement of S H Rz(π/4) H|0> in computational basis: \n", np.abs(S_H_Rz_H_zero)**2)

Rz_T_X_one = circuit_sequence(one, [X, T, Rz(np.pi/4)])
print("Rz(π/4) T X|1> =\n", Rz_T_X_one)
print("Measurement of Rz(π/4) T X|1> in computational basis: \n", np.abs(Rz_T_X_one)**2)

X_Y_Z_S_T_Rx_Ry_Rz_zero = circuit_sequence(zero, [Rz(np.pi/4), Ry(np.pi/4), Rx(np.pi/4), T, S, Z, Y, X])
print("X Y Z S T Rx(π/4) Ry(π/4) Rz(π/4)|0> =\n", X_Y_Z_S_T_Rx_Ry_Rz_zero)
print("Measurement of X Y Z S T Rx(π/4) Ry(π/4) Rz(π/4)|0> in computational basis: \n", np.abs(X_Y_Z_S_T_Rx_Ry_Rz_zero)**2)
```

Normalization check:

```python
def is_normalized(state):
    probabilities = np.abs(state.flatten()) ** 2
    return np.isclose(np.sum(probabilities), 1.0)

print(is_normalized(S_H_Rz_H_zero))
print(is_normalized(Rz_T_X_one))
print(is_normalized(X_Y_Z_S_T_Rx_Ry_Rz_zero))
```
