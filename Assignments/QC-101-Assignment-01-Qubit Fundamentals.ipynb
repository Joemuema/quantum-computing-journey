Exercise 1:

```python

import numpy as np

# Define the computational basis states
zero = np.array([1, 0], dtype=complex)
one = np.array([0, 1], dtype=complex)

# Define an equal superposition state
alpha = np.sqrt(3) / 2
beta = 1 / 2

qubit = alpha * zero + beta * one

print("Qubit state:")
print(qubit)

# Calculate measurement probabilities
prob_zero = abs(qubit[0]) ** 2
prob_one = abs(qubit[1]) ** 2

print("\nMeasurement probabilities:")
print("P(0):", prob_zero)
print("P(1):", prob_one)

# Verify normalization
print("\nTotal probability:", prob_zero + prob_one)
```

Exercise 2:

```python

import numpy as np

# Equal superposition state
qubit = np.array([
    np.sqrt(3) / 2,
    1 / 2
])

# Calculate measurement probabilities
probabilities = np.abs(qubit) ** 2

# Perform 1000 independent measurements
results = np.random.choice(
    [0, 1],
    size=10000,
    p=probabilities
)

# Count measurement outcomes
counts = {
    0: np.count_nonzero(results == 0),
    1: np.count_nonzero(results == 1)
}

print("Measurement results:", counts)
print("Probability of measuring 0:", counts[0] / 10000)
print("Probability of measuring 1:", counts[1] / 10000)
```

Exercise 3:

```python

import numpy as np

# Hadamard gate
H = (1 / np.sqrt(2)) * np.array([
    [1, 1],
    [1, -1]
])

# Initial state |1>
one = np.array([0, 1], dtype=complex)

# Apply Hadamard gate
new_state = H @ one

print("State after Hadamard:")
print(new_state)

# Calculate probabilities
probabilities = np.abs(new_state) ** 2

print("Measurement probabilities:")
print(probabilities)
```

Execise 4: Why is a quantum superposition different from a classical computer randomly choosing between 0 and 1?

Quantum superposition is not just a state of randomly being either a 0 or 1, but is the state of a qubit having different complex probabilities of being a 0 or 1 until measured, of which these probabilities when expressed as waves can interfere with each other since they occur simultaneously.
